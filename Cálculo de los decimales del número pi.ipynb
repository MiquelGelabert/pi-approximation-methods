# <div id='id2' /> BREVE HISTORIA

Desde el antiguo Egipto hasta la actualidad, pasando por la Gracia clásica y el renacimiento europeo, la curiosidad por saber el enésimo decimal del número pi ha estado presente. El número pi se define como la relación entre la longitud y el diámetro de una circunferencia. Además, es un número irracional y trascendente, lo cual significa que tiene infinitos números decimales que no siguen ningún patrón y que no es raíz de ninguna ecuación algebraica con coeficientes enteros no todos nulos, respectivamente. Es uno de los números más importantes de las matemáticas, entre otros motivos, porque aparece en gran parte de sus ramas, como el cálculo o la probabilidad, aunque también aparece prácticamente en todas las áreas de la física, como por ejemplo en la gravitación o en electromagnetismo.

Una de las primeras estimaciones se realizó en el antiguo Egipto, cerca del 1800 a.C., y se calculaba como 256/81≈3.16049, aunque también, en esa época, en Mesopotamia se aproximaba como 3+(1/8)=3.125. También, la Biblia, en Reyes (I, 7, 23), sitúa pi como 3. Después, en la antigua Grecia, Arquímedes, siglo III a.C., desarrolla un método (basado en la exhausción de polígonos) con el que obtiene una cota del valor de pi entre 3.1408 y 3.1428 y Ptolomeo, siglo II d.C., lo aproximó como 377/120≈3.1416, siguiendo el método de Arquímedes.

En la antigua China, sobre el año 120, Zhang Heng fue pionero en usar la aproximación raíz cuadrada de 10, que es aproximadamente 3.1622. Un siglo después, Wang Fang utilizó la aproximación 142/45≈3.155 y Liu Hui 3,14159. A finales del siglo V, el matemático Zu Chongzhi dio con dos aproximaciones racionales: 22/7≈3.142857 y 355/113≈3.141592, la segunda siendo una muy buena aproximación para la época, tanto fue así que no se superó hasta nueve siglos más tarde.

El matemático indio Aryabhata estimó en el mismo siglo V pi como 3.1416, aunque mucho antes, nueve siglos antes, en la India, se aproximaba 339/108 = 3.1388 en los Shatapatha Brahmana. Dos siglos después, Brahmagupta, uno de los matemáticos indios más importantes, usa la misma aproximación de Zhang Heng. Hacia 1400, Madhava fue el primero en usar series, obteniendo 10 dígitos, viéndose así como las series iban a ser útiles en esta búsqueda.

En el siglo IX, Al-Jwarizmi, en su Algebra, escribe que el hombre práctico usa 22/7, el geómetra usa 3 y el astrónomo usa 3.1416. El matemático persa Ghiyath al-Kashi, con una base sexagesimal, fue capaz de calcular nueve dígitos, lo que equivale a una aproximación de 16 dígitos decimales.

Empezando el renacimiento europeo, otros matemáticos siguieron añadiendo más lados al método de Arquímedes. Por ejemplo, el famoso matemático Fibonacci (siglo XII-XIII) también se animó a dar su aproximación estrechando el intervalo, Viète (siglo XII) con 393216 lados consiguió 9 decimales exactos y en 1593 el flamenco Adriaan van Roomen obtiene una precisión de 16 decimales.

A partir de la Edad Moderna, la cantidad de decimales exactos calculados fue creciendo exponencialmente gracias al desarrollo del cálculo infinitesimal de Newton y Leibniz, el uso de series y de las relaciones trigonométricas. En 1610, el matemático Ludolph van Ceulen calculó los 35 primeros decimales de pi, los cuales fueron igualados por Snell once años después. En 1630, Christoph Grienberger, un astrónomo austriaco, batió el récord de dígitos, llegando hasta los 39. John Wallis desarrolló su famoso Producto de Wallis en 1655. En 1665 sir Isaac Newton desarrolla la serie arcsin, aunque no obtuvo un avance en la cantidad de decimales, pero fue el precursor de la utilización de este tipo de series como también de usar el cálculo diferencial. Leibniz desarrolló una serie, la cual lleva su nombre, en 1682. En 1699 el matemático inglés Abraham Sharp calculó pi con una precisión de 71 dígitos decimales usando la serie de arctan a petición de Edmond Halley. El inglés William Oughtred fue el primero que usó la letra griega π como símbolo de la relación entre la circunferencia y su diámetro. Fue en el año 1706 cuando el galés William Jones afirmó que π=3.14159... y propuso usar siempre el símbolo π. El gran matemático Leonhard Euler fue el que, al adoptarlo en 1737, lo convirtió en la notación habitual que se usa hasta nuestros días. El matemático japonés Takebe, en el año 1722, con el método de Arquímedes, consiguió 41 decimales. Entre 1760 y 1800 se produjeron en paralelo la demostración de la irracionalidad de pi gracias a Lambert y de la trascendencia de pi a Joseph Liouville, ya prevista por Euler años antes. En 1789 el matemático esloveno Jurij Vega, mediante la fórmula de John Machin (se basaba en aproximaciones de arcotangentes), descubierta en 1706, fue el primero en averiguar los primeros 126 decimales; este récord se mantuvo durante 52 años, hasta que en 1841 William Rutherford calculó 152 y en 1853 se superó con 440. El matemático inglés William Shanks trabajó, durante 20 años, en hallar los dígitos decimales de π, habiendo obtenido 707 decimales en 1873. En 1948, Ferguson recalculó π con 808 decimales con la ayuda de una calculadora electrónica. # <div id='id4' /> 1. Método de Arquímedes

Arquímedes de Siracusa, siglo III a.C., matemático y físico de la antigua Grecia, desarrolló un método de exhausción, ideado por Eudoxo de Cnido, para determinar el valor del número pi, uno de los primeros de los que se tiene constancia y que, además, es considerablemente preciso para su época. Aunque, una vez que se dominó la trigonometría, se recogió el método de Arquímedes de exhausción actualizándolo con trigonometría.

## Original

Arquímedes desarrolló un algoritmo que se basa en una fórmula recursiva que comparara los perímetros de los polígonos de n lados inscritos y circunscritos en una circunferencia de radio r. La longitud de la circunferencia que queda delimitada entre los dos polígonos estará acotada entre estos dos perímetros. Empezando con unos valores inciales, para un polígono de 3 lados (un triángulo):
### $$a_0=\sqrt{12}; b_0=3.$$ 
Se itera el siguiente algoritmo:
### $$a_{n+1}=\frac{2a_nb_n}{a_n+b_n}$$
###  $$b_{n+1}=\sqrt{a_{n+1}b_n}$$
Así acotando el valor de pi de la siguiente manera:
###  $$a_n<\pi<b_n$$
Donde tanto $a_n=\frac{p_n}{2r}$ como $b_n=\frac{P_n}{2r}$, en los que $p_n$ y $P_n$ son los perímetros de los polígonos inscritos y circunscritos respectivamente.

```python
fig, axs = plt.subplots(1, 3, figsize=(12,4))
for ax in axs:
    ax.set_aspect('equal')
    ax.set_xlim(-1.2, 1.2)
    ax.set_ylim(-1.2, 1.2)
    ax.axis('off')

radius = 1
circ = plt.Circle((0, 0), radius, fill=False, lw=2)
axs[0].add_artist(circ)
axs[0].set_title('Polígono 6 lados')

circ = plt.Circle((0, 0), radius, fill=False, lw=2)
axs[1].add_artist(circ)
axs[1].set_title('Polígono 12 lados')

circ = plt.Circle((0, 0), radius, fill=False, lw=2)
axs[2].add_artist(circ)
axs[2].set_title('Polígono 96 lados')

for i, n in enumerate([6, 12, 96]):
    polygon_points = np.array([(radius*np.cos(j*2*np.pi/n), radius*np.sin(j*2*np.pi/n)) for j in range(n)])
    axs[i].plot(polygon_points[:,0], polygon_points[:,1], lw=2, color='blue')
    axs[i].plot([polygon_points[-1,0], polygon_points[0,0]], [polygon_points[-1,1], polygon_points[0,1]], lw=2, color='blue')
    
    r_circum = radius / np.cos(np.pi/n) 
    circum_points = np.array([(r_circum*np.cos(j*2*np.pi/n), r_circum*np.sin(j*2*np.pi/n)) for j in range(n)])
    axs[i].plot(circum_points[:,0], circum_points[:,1], lw=2, color='red')
    axs[i].plot([circum_points[-1,0], circum_points[0,0]], [circum_points[-1,1], circum_points[0,1]], lw=2, color='red')
plt.show()
```

```python
def arquimedes(N):
    a,b,n,menor,mayor,pi = 12**(1/2),3,[],[],[],[]
    for k in range(3,N+1):
        n.append(k)
        a = 2*a*b/(a+b)
        b = (a*b)**(1/2)
        menor.append(b)
        mayor.append(a)
        pi.append("< 𝜋 <")
    tabla = pd.DataFrame(list(zip(n,menor,pi,mayor)), columns=['n', 'a', '𝜋', 'b'])
    pd.set_option('display.precision', 70)
    return tabla
arquimedes(100)
```

Como se puede ver, para las primeras iteraciones, con n=3, 4, 5... se consigue ya una buena aproximación. Para n=100, se tiene acotado pi como mínimo hasta el sexto decimal. Ahora bien, este método depende de los valores iniciales que, en este caso, han sido elegidos convenientemente.

## Trigonometría

Siguiendo con el método de exhausción de los polígonos y añadiendo trigonometría, se puede calcular el lado de un polígono de n lados inscrito en una circunferencia de radio r de la siguiente manera:
### $$l=2\sin{\left ( \frac{180º}{n}\right )}r$$
Y el lado de un polígono de n lados circunscrito en una circunferencia de radio r:
### $$L=l\sec{\left ( \frac{180º}{n}\right )}=2\tan{\left ( \frac{180º}{n}\right )}r$$
Por lo tanto, los perímetros del polígono inscrito y del circunscrito son, respectivamente:
### $$p_n=2n\sin{\left ( \frac{180º}{n}\right )}r=nl$$
### $$P_n=2n\tan{\left ( \frac{180º}{n}\right )}r=nL$$
Es así como se puede acotar la longitud de la circunferencia superiormente con el perímetro del circunscrito e inferiormente con el perímetro del inscrito:
### $$p_n<2\pi r<P_n$$
Por lo que se puede, dividiendo entre $2r$, acotar pi de la siguiente manera:
### $$n\sin{\left ( \frac{180º}{n}\right )}<\pi<n\tan{\left ( \frac{180º}{n}\right )}$$

```python
def trig_arq(N):
    n,menor,mayor,pi=[],[],[],[]
    for k in range(1,N+1):
        n.append(k)
        menor.append(k*np.sin(radians(180)/k))
        mayor.append(k*np.tan(radians(180)/k))
        pi.append("< 𝜋 <")
    tabla = pd.DataFrame(list(zip(n,menor,pi,mayor)), columns=['n', 'menor', '𝜋', 'mayor'])
    pd.set_option('display.precision', 70)
    return tabla
trig_arq(100)
```

Aunque sea un método más avanzado que el original, no es igual de eficiente, ya que para n = 100, solo obtiene 3 decimales.

## Comparación de la convergencia con los métodos de Arquímedes

```python
def comp_conv_arq(N):
    n,menor1,mayor1,a,b,pi,menor2,mayor2 = [],[],[],12**(1/2),3,[],[],[]
    for k in range(3,N+1):
        n.append(k)
        a = 2*a*b/(a+b)
        b = (a*b)**(1/2)
        menor1.append(a)
        mayor1.append(b)
        menor2.append(k*np.sin(np.pi/k))
        mayor2.append(k*np.tan(np.pi/k))
        pi.append(np.pi)
    tabla = pd.DataFrame(list(zip(n,menor1,pi,mayor1,menor2,pi,mayor2)), columns=['n', 'menor sin', '𝜋', 'mayor sin','menor con', 'pi', 'mayor con'])
    pd.set_option('display.precision', 19)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(3,N+1), mayor1, label="Pi sobreestimado sin trigonometria",color="b")
    ax.plot(range(3,N+1), menor1, label="Pi subestimado sin trigonometria",color="b")
    ax.plot(range(3,N+1), mayor2, label="Pi sobreestimado con trigonometria",color="r")
    ax.plot(range(3,N+1), menor2, label="Pi subestimado con trigonometria",color="r")
    ax.plot(range(3,N+1), pi, "--", label="Pi real", linewidth=2, color="green")
    plt.title('Comparación de la convergencia de Pi respecto n con los métodos de Arquímedes')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
    return tabla
comp_conv_arq(100)
```

El método de Arquímedes sin trigonometría es mucho más rápido en cuanto a precisión que con trigonometría. Además, cabe notar que, con trigonometría, pi subestimado se acerca más a pi que pi sobreestimado. La ventaja de usar trigonometría es que solo de la cantidad de lados n y no de ningún valor inicial.

# <div id='id5' /> 2. Métodos probabilísticos

Aunque no lo parezca, el número pi aparece en la probabilidad. Ahora bien, el método de Montecarlo y el método de Buffon aparecen de manera indirecta en la probabilidad, uno por la geometría y otro por la trigonometría.

## Método de Montecarlo

El método de Montecarlo consiste en inscribir una circunferencia en un cuadrado y calcular la proporción entre el área de uno con el área del otro. Si consideramos una circunferencia de radio r, la relación de áreas, que le denominaremos "k", permanece constante y se puede escribir como:
### $$k=\frac{A_{cuadrado}}{A_{circunferencia}}=\frac{(2r)^2}{\pi r^2}=\frac{4}{\pi}$$
Por tanto: 
### $$\pi=\frac{4}{k}$$
Entonces, si se calcula de manera indirecta/aproximada esta relación de áreas, se puede calcular el valor aproximado del número pi.
    
Se generarán puntos aleatorios en el cuadrado y se contarán cuántos están dentro de la circunferencia y esa cantidad será su área aproximada y la cantidad total de puntos será el área aproximada del cuadrado, por lo que su cociente resultará k.
El algoritmo genera puntos aleatorios y cuenta los que estén a una distancia menor de una unidad (el radio arbitrario que se ha elegido). Después calculará el cociente entre 4 y k, o dicho de otra manera: pi será cuatro veces el cociente entre la cantidad de puntos dentro y la cantidad total de puntos.
### $$\pi=\frac{4}{\frac{N}{a}}=4\frac{a}{N}$$
Donde "a" es la cantidad de puntos dentro y N la cantidad total de puntos.

```python
def montecarlo(N):
    a,b,c,d,f,g = 0,0,[],[],[],[]
    for i in range(1,N+1):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        if x**2+y**2<=1:
            a = a + 1
            c.append(x)
            d.append(y)
        else:
            b = b +1
            f.append(x)
            g.append(y)
    return (4*a)/N
montecarlo(100000)
```

```python
def montecarlo_plot(N):
    a,b,c,d,f,g = 0,0,[],[],[],[]
    for i in range(1,N+1):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        if x**2+y**2<=1:
            a = a + 1
            c.append(x)
            d.append(y)
        else:
            b = b +1
            f.append(x)
            g.append(y)
    print("El número de puntos dentro de la circunferencia es ", a)
    print("El número de puntos dentro del cuadrado es ", a+b)
    pi = (4*a)/N
    print("La aproximación al número pi calculada con el Método Montecarlo con", N, "puntos es", pi)
    fig,ax = plt.subplots(figsize=(10,10))
    ax.scatter(c,d,s=2, color="g")
    ax.scatter(f,g,s=2, color="r")
    ax.axis("off")
montecarlo_plot(100000)
```

Como se observa, los puntos verdes están a una distancia inferior a una unidad y los que están más alejados de esa unidad están representados en rojo. En total hay 100000 puntos, todos ellos están dentro del cuadrado y solo unos cuantos están dentro, en este caso: 78597. Calculando el cociente entre $4\frac{a}{N}$, donde 'a' es la cantidad de puntos dentro, se debería de obtener la aproximación de pi con nuestro método. En esta iteración se obtuvo la aproximación de pi 3.14388. Normalmente, con N = 100000, se obtiene una precisión de 2 decimales, un método poco eficaz en este sentido.

```python
def error_montecarlo(N):
    error=[]
    for n in range(1,N+1):
        error.append(abs(np.pi-montecarlo(n)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(1,N+1),error, color="r",label="Error Montecarlo")
    plt.title("Error de Montecarlo en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_montecarlo(1000)
```

## Método de Buffon

El método de Buffon fue planteado por Georges Louis Leclerc, conde de Buffon. Tiene diferentes casos, aunque en este notebook se tratará el caso en el que se lanza una aguja en un papel, la cual es igual de larga que la distancia entre las rectas paralelas (por eso basta una recta, con la que mediremos la cantidad de intersecciones). Se puede demostrar que la probabilidad es proporcional a $\frac{2}{\pi}$ y, por tanto, pi se puede aproximar como:
### $$\pi=\frac{2N}{A}$$
Donde N es el número total de intentos y A el número de veces que alguna aguja cruza la recta.

Cabe decir que en el caso en el que no se dé ninguna intersección con la recta, pi no se podría calcular con este método debido a una indefinición. Sí que es verdad que no acaba de ser fidedigno con la realidad (puede pasar que ninguna se cruce), pero se ha decidido que, para que los gráficos se mantengan continuos, se ha determinado 0.00000000001 como la cantidad de veces que una aguja ha cruzado la recta cuando, realmente, no ha habido ninguna.

```python
def buffon(N):
    L, pi, crosses = 1, 0, 0
    for i in range(1, N+1):
        x = random.random() * L
        y = random.random() * L
        theta = random.vonmisesvariate(0, 0)
        x_end = x + (L*cos(theta))
        y_end = y + (L*sin(theta))
        if x < (L/2) * abs(cos(theta)) or x > L - (L/2) * abs(cos(theta)):
            crosses += 1
    if crosses != 0:
        return  (2*L*N) / ((crosses) * L)
    if crosses == 0: 
        return (2*L*N) / ((0.00000000001) * L)
buffon(100)
```

```python
def buffon_plot(N):
    L, pi, crosses = 1, 0, 0
    fig,ax = plt.subplots(figsize=(10,8))
    for i in range(1, N+1):
        x = random.random() * L
        y = random.random() * L
        theta = random.vonmisesvariate(0, 0)
        x_end = x + (L*cos(theta))
        y_end = y + (L*sin(theta))
        if x < (L/2) * abs(cos(theta)) or x > L - (L/2) * abs(cos(theta)):
            crosses += 1
        ax.plot([x, x_end], [y, y_end], 'grey', lw=1)
    ax.plot([L/2, L/2], [-1, 2], 'k-', lw=1)
    plt.xlim(-L, L*2)
    plt.ylim(-L, L*2)
    plt.show()
buffon_plot(20)
```

El error va disminuyendo cuantas más agujas tiradas, pero no es un método muy preciso, ya que, para 1000 agujas, sigue teniendo un error de 0.2.

# <div id='id6' /> 3. Métodos con series

Uno de los métodos más efectivos en toda la historia de la búsqueda del número pi han sido las series. Por lo que, cuántos más términos se le añadan a la suma, más dígitos se obtendrán. 

Aunque quizás surge la duda siguiente: ¿puede una suma infinita tener un valor exacto? O también: ¿puede una serie resultar el número pi? Cuando una serie resulta un número finito (que no resulte infinito, que es lo normal) se dice que la serie es convergente y que converge en ese número finito. Normalmente, las series son divergentes, que no convergen, pero, entre las que sí son convergentes, muy pocas lo hacen en el número pi. Para saber si una serie es convergente, se pueden aplicar una serie de criterios y para encontrar una serie que converja en pi, seguramente, se tendrán que usar relaciones trigonométricas. Se presentarán aquellas series más famosas que les sucede lo explicado anteriormente.

La mayoría de estas series que se han utilizado provienen de la trigonometría y de la geometría. Se pudieron obtener unas expresiones que describían una suma de distancias que convergen en pi o en una expresión que depende de él, la fórmula de Leibniz o el algoritmo de Newton son dos claros ejemplos porque provienen de arcosenos y arcotangentes, respectivamente. Existen una gran cantidad de series más que las presentadas, pero muchas de ellas son equivalentes entre ellas o entre las siguientes. Sobre todo, la fórmula de Leibniz tiene muchas variantes, aunque se ha escogido por solo representar la más común. Se ha decidido incluir el producto de Wallis que, como el propio nombre indica, es un producto, no una serie, pero, como la única diferencia es que, en lugar de sumarlos, los términos se multiplican, se presentará en esta sección. ## Fórmula de Leibniz 
### $$\arctan(1)=4\sum_{n=1}^{\infty} \frac{(-1)^n}{2n+1}=\pi$$

```python
def leibniz(N): 
    sum=0
    for n in range(0,N+1):
        sum += ((-1)**(n))/(2*n+1)
    return 4*sum
leibniz(100)
```

```python
def conv_leibniz(N):
    pi=[]
    pi_real=[]
    for n in range(0,N+1):
        pi.append(leibniz(n))
        pi_real.append(np.pi)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1), pi, label="Pi estimado", color="magenta")
    ax.plot(range(0,N+1), pi_real, "--", color="green", label="Pi real")
    plt.title('Convergencia de Pi respecto n con la fórmula de Leibniz')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
conv_leibniz(100)
```

```python
def error_leibniz(N):
    error=[]
    for p in range(0,N+1):
        error.append(abs(np.pi-leibniz(p)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1),error, color="magenta", label="Error Leibniz")
    plt.title("Error de Leibniz en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_leibniz(100)
```

El método de Leibniz sigue una serie alternada, por lo que a cada término cambia entre sumar y restar. Es así como a cada término o sobreestima o infraestima el valor de pi. Sin embargo, como se observa en las gráficas, va corrigiendo el error, hasta llegar a ser bastante preciso.

## Fórmula de Euler 
### $$\sum_{n=1}^{\infty}\frac{1}{n^2}=\frac{\pi^2}{6}$$

```python
def euler(N): 
    sum=0
    for n in range(1,N+1):
        sum += 1/n**2
    return (6*sum)**0.5
euler(100)
```

## Algoritmo de Newton 
### $$\sum_{n=0}^{\infty} \frac{\binom{2n}{n}}{(2n+1)16^n}=\frac{\pi}{3}$$

```python
def newton(N): 
    sum=0
    for n in range(0,N+1):
        sum += (factorial(2*n)/(factorial(n)*factorial(2*n-n)))/((2*n+1)*16**n)
    return 3*sum
newton(100)
```

```python
def conv_newton(N):
    pi=[]
    pi_real=[]
    for n in range(0,N+1):
        pi.append(newton(n))
        pi_real.append(np.pi)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1), pi, label="Pi estimado", color="y")
    ax.plot(range(0,N+1), pi_real, "--", color="green", label="Pi real", linewidth=1)
    plt.title('Convergencia de Pi respecto n con el algoritmo de Newton')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
conv_newton(100)
```

```python
def error_newton(N):
    error=[]
    for p in range(0,N+1):
        error.append(abs(np.pi-newton(p)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1),error, color="y", label="Error Newton")
    plt.title("Error de Newton en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_newton(100)
```

## Producto de Wallis 
### $$\prod_{n=1}^{\infty}\left ( \frac{2n}{2n-1}\cdot \frac{2n}{2n+1} \right )=\frac{\pi}{2}$$

```python
def wallis(N):
    p=1
    for n in range(1,N+1):
        p *= (2*n/(2*n-1))*(2*n/(2*n+1))
    return 2*p
wallis(10000)
```

```python
def conv_wallis(N):
    pi=[]
    pi_real=[]
    for n in range(1,N+1):
        pi.append(wallis(n))
        pi_real.append(np.pi)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(1,N+1), pi, label="Pi estimado", color="brown")
    ax.plot(range(1,N+1), pi_real, "--", color="green",label="Pi real", linewidth=1)
    plt.title('Convergencia de Pi respecto n con el producto de Wallis')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
conv_wallis(100)
```

```python
def error_wallis(N):
    error=[]
    for p in range(0,N+1):
        error.append(abs(np.pi-wallis(p)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1),error, color="brown", label="Error Wallis")
    plt.title("Error de Wallis en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_wallis(100)
```

El producto de Wallis es un método que va multiplicando términos cada vez más pequeños, por lo que nunca sobrepasa el valor real de pi, siempre lo subestima, aunque va reduciendo el error de manera similar que la fórmula de Euler.

## Fórmula Bailey-Borwein-Plouffe 
### $$\sum_{n=0}^{\infty}\frac{1}{16^n}\left ( \frac{4}{8n+1}-\frac{2}{8n+4}-\frac{1}{8n+5}-\frac{1}{8n+6} \right )=\pi$$

```python
def bbp(N):
    sum = 0
    for n in range(0,N+1):
        sum += (1/16**n)*((4/(8*n+1))-(2/(8*n+4))-(1/(8*n+5))-(1/(8*n+6)))
    return sum
bbp(100)
```

```python
def error_bbp(N):
    error=[]
    for p in range(0,N+1):
        error.append(abs(np.pi-bbp(p)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1),error, color="red", label="Error BBP")
    plt.title("Error de BBP en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_bbp(100)
```

La fórmula de Bailey-Borwein-Plouffe es una serie ya mucho más compleja que las anteriores, tiene muchos más términos, aunque siempre genera valores menores al real. Estos términos consiguen una precisión ampliamente superior, reduciendo drásticamente el error.

## Algoritmo de Chudnovsky 
### $$12\sum_{n=0}^{\infty} \frac{(-1)^n(6n)!(13591409+545140134n)}{(3n)!(n!)^3(640320)^{3n+\frac{3}{2}}}=\frac{1}{\pi}$$

## Serie de Ramanujan-Sato 
### $$\frac{2\sqrt{2}}{99^2}\sum_{n=0}^{\infty} \frac{(4n)!}{n!^4}\frac{26390n+1103}{396^{4n}}=\frac{1}{\pi}$$

```python
def ramanujan(N):
    sum = 0
    for n in range(0,N+1):
        sum += (factorial(4*n)*(26390*n+1103))/(((factorial(n))**4)*(396)**(4*n))
    return (99**2)/(2*((2)**(1/2))*sum)
ramanujan(100)
```

```python
def error_ramanujan(N):
    error=[]
    for p in range(0,N+1):
        error.append(abs(np.pi-ramanujan(p)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1),error, color="orange", label="Error Ramanujan")
    plt.title("Error de Ramanujan en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_ramanujan(100)
```

## Fórmula de Nilakantha 
### $$3+4\sum_{n=0}^{\infty} \frac{(-1)^n}{(2n+3)^3-(2n+3)}=\pi$$

```python
def nilakantha(N):
    sum = 0
    iteracion = 0
    for n in range(2,N+1,2):
        iteracion += 1
        if iteracion%2 == 0:
            sum -= 1/(((n))*(n+1)*(n+2))
        else:
            sum += 1/(((n))*(n+1)*(n+2))
    return 3+4*sum
nilakantha(1000)
```

```python
def conv_nilakantha(N):
    pi=[]
    pi_real=[]
    for n in range(2,N+1):
        pi.append(nilakantha(n))
        pi_real.append(np.pi)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(2,N+1), pi, label="Pi estimado", color="purple")
    ax.plot(range(2,N+1), pi_real, "--", color="green",label="Pi real", linewidth=1)
    plt.title('Convergencia de Pi respecto n con Nilakantha')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
conv_nilakantha(100)
```

```python
def error_nilakantha(N):
    error=[]
    for p in range(2,N+1):
        error.append(abs(np.pi-nilakantha(p)))
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(2,N+1),error, color="purple", label="Error Nilakantha")
    plt.title("Error de Nilakantha en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_nilakantha(100)
```

## Serie de Ibo 
### $$8\sum_{n=1}^{\infty} \frac{1}{4n(4n+2)}=\pi,n=4, 8, 12, 16, 20...$$

```python
def ibo(N):
    sum = 0
    for n in range(1, N+1, 4):
        sum += 8/(n*(n+2))
    return sum
ibo(100)
```

## Función Zeta de Riemann 
### $$\zeta(2)=\sum_{n=1}^{\infty} \frac{1}{n^2}=\frac{\pi^2}{6}$$ $$\zeta(4)=\sum_{n=1}^{\infty} \frac{1}{n^4}=\frac{\pi^4}{90}$$ $$\zeta(6)=\sum_{n=1}^{\infty} \frac{1}{n^6}=\frac{\pi^6}{945}$$ $$\zeta(8)=\sum_{n=1}^{\infty} \frac{1}{n^8}=\frac{\pi^8}{9450}$$

```python
def zeta(s, N):
    sum = 0
    for n in range(1, N+1):
        sum += 1/n**s
    if s == 2:
        return (6*sum)**(1/s)
    if s == 4: 
        return (90*sum)**(1/s)
    if s == 6:
        return (945*sum)**(1/s)
    if s == 8: 
        return (9450*sum)**(1/s)
```

## Números catalanes 
### $$C_{n}=\frac{\binom{2n}{n}}{n+1}\approx\frac{4^n}{n^{3/2}\sqrt\pi}\Rightarrow \frac{4^{2n}(n+1)^2}{n^3\binom{2n}{n}^2}=\pi$$

## Fórmula de Stirling 
### $$n!\approx \sqrt{2\pi n}\binom{n}{e}^n\Rightarrow \frac{n!^2}{2n\binom{n}{e}^{2n}}=\pi$$

## Algoritmo de Gauss-Legendre

Se establecen los valores iniciales:

### $$a_0=1;  b_0=\frac{1}{\sqrt2}; t_0=\frac{1}{4}; p_0=1$$

Iterar:

### $$x_{n+1}=\frac{a_n+b_n}{2} $$
### $$y_{n+1}=\sqrt{a_nb_n} $$
### $$t_{n+1}=t_n-p_n(a_n-x_{n+1})^2$$
### $$a_{n+1}=x_{n+1}; b_{n+1}=y_{n+1}; p_{n+1}=2p_n$$

Se calcula pi aproximado:

### $$\pi=\frac{(a_n+b_n)^2}{4t_n} $$

```python
def gauss(N):
    a,b,t,p = 1, 1/((2)**(1/2)), 1/4, 1
    for n in range(0,N+1):
        x = (a+b)/2
        y = (a*b)**(1/2)
        t = t - p*(a-x)**2
        a = x
        b = y
        p = 2*p
    return ((a+b)**2)/(4*t)
gauss(100)
```

```python
def conv_gauss(N):
    pi=[]
    pi_real=[]
    for n in range(0,N+1):
        pi.append(gauss(n))
        pi_real.append(np.pi)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(0,N+1), pi, label="Pi estimado", color="purple")
    ax.plot(range(0,N+1), pi_real, "--", color="green",label="Pi real", linewidth=1)
    plt.title('Convergencia de Pi respecto n con Gauss')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
conv_gauss(20)
```

```python
def error_gauss(N):
    error=[]
    n=[]
    for p in range(1,N+1):
        error.append(abs(np.pi-gauss(p)))
        n.append(p)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(n,error, color="purple", label="Error Gauss")
    plt.title("Error de Gauss en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_gauss(100)
```

El algoritmo de Gauss-Legendre es un método extremadamente preciso, aunque es muy sensible a los valores iniciales. Se han elegido los óptimos para que se pueda sacar la mejor versión del método, pero si se cambian se vería como el error difiere significativamente.

## Algoritmo de Borwein-Borwein

Se establece los valores iniciales:

### $$s_0=\sqrt2 -1;  t_0=6-4\sqrt2$$

Iterar:

### $$s_{n+1}=\frac{1-(1-s_n^4)^{1/4}}{1+(1-s_n^4)^{1/4}}$$
### $$t_{n+1}=t_n(1+s_{n+1}^4)-2^{2n+3}s_{n+1}(1+s_{n+1}+s_{n+1}^2)$$

Se calcula pi aproximado:

### $$\pi=\frac{1}{t_n} $$

```python
def borwein(N):
    s = 2**(1/2)-1
    t = 6-4*(2)**(1/2)
    for n in range(1,N+1):
        s = (1-(1-s**4)**(1/4))/(1+(1-s**4)**(1/4))
        t = t*(1+s)**4-2*(4**n)*s*(1+s+s**2)
    return 1/t
borwein(100)
```

```python
def conv_borwein(N):
    pi=[]
    pi_real=[]
    for n in range(1,N+1):
        pi.append(borwein(n))
        pi_real.append(np.pi)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(range(1,N+1), pi, label="Pi estimado", color="yellow")
    ax.plot(range(1,N+1), pi_real, "--", color="green",label="Pi real", linewidth=1)
    plt.title('Convergencia de Pi respecto n con Borwein')
    plt.xlabel('n')
    plt.ylabel('Pi')
    ax.legend();  
conv_borwein(100)
```

```python
def error_borwein(N):
    error=[]
    n=[]
    for p in range(1,N+1):
        error.append(abs(np.pi-borwein(p)))
        n.append(p)
    fig,ax = plt.subplots(figsize=(10,8))
    ax.plot(n,error, color="yellow", label="Error Borwein")
    plt.title("Error de Borwein en calcular el número pi respecto n")
    plt.xlabel('n')
    plt.ylabel('Error')
error_borwein(100)
```

El método de Borwein-Borwein es un método iterativo, que también depende de los valores iniciales. Es uno de los métodos iterativos más precisos. El primer error se encuentra en el noveno decimal.

### 1.1. Integrales
Las integrales fueron un gran avance que desarrollaron Newton y Leibniz. Entre las siguientes, se encuentran algunas que ambos usaron para encontrar el valor de pi.

### $$\int_{0}^{1} \frac{dx}{1+x^2}=\frac{\pi}{4}$$
### $$\int_{-\infty}^{\infty} \frac{dx}{1-x^2}=\pi$$
### $$\int_{0}^{\infty} \frac{\sin{x}}{x}dx=\frac{\pi}{2}$$
### $$\int_{-\infty}^{\infty} e^{-x^2}dx=\sqrt{\pi}$$

Esta última se utiliza a la hora de resolver la integral de la campana de Gauss. Por lo que, en probabilidad, en la distribución normal, aparece pi de la siguiente manera:
### $$G_{\mu, \sigma^2} (x) = \int_{-\infty}^{x} \phi_{\mu, \sigma^2} (z) dz = \frac{1}{\sigma\sqrt{2\pi}} \int_{-\infty}^{x} e^{-\frac{(z-\mu)^2}{2\sigma^2}} dz$$

Incluso en un reto sin una solución final, nos adentramos con todo nuestro ingenio y capacidad para aportar el máximo y mejorar...
